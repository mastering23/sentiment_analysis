```python
# Import SpaCy and load the language library . Remember to use a larger model!
import spacy
nlp = spacy.load('en_core_web_lg')
```

```python
tokens = nlp(u'food apple orange hotdog')
```

```python
for t1 in tokens:
    for t2 in tokens:
        print(t1.text,t2.text,t1.similarity(t2))
```

```python
# import spatial and define a cosine_similarity function
from scipy import spatial
cosine_similarity = lambda v1,v2: 1-spatial.distance.cosine(v1,v2)
```

```python
# Write an expression for vector arithmetic
# For Example: new_vector = word1- word2 + word3
car = nlp("car").vector
boat = nlp("boat").vector
airplane = nlp("airplane").vector
new_vector = car - boat + airplane
```

```python
# List the top ten closest vectors in the vocabulary to the result of the expression above
comp_similarities = []
for s in nlp.vocab.vectors:  
    _ = nlp.vocab[s]

for word in nlp.vocab:  
    if word.has_vector:  
        if word.is_lower:  
            if word.is_alpha:  
                similarity = cosine_similarity(new_vector, word.vector)
                comp_similarities.append((word, similarity))  
```

```python
comp_similarities = sorted(comp_similarities, key=lambda item: -item[1])
```

```python
print([t[0].text if hasattr(t[0], 'text') else str(t[0]) for t in comp_similarities[:10]])
```

<strong>Challenge: Write a function that takes in 3 strings, performs a-b+c arithmetic and returns a top-ten result

```python
def vector_match(a, b, c):
    vector_a = nlp(a).vector
    vector_b = nlp(b).vector
    vector_c = nlp(c).vector
    new_vector = vector_a - vector_b + vector_c
    
    comp_similarities = [] 

    for word in nlp.vocab:  
        if word.has_vector and word.is_lower and word.is_alpha:
            similarity = cosine_similarity(new_vector, word.vector)
            comp_similarities.append((word.text, similarity))  

    comp_similarities = sorted(comp_similarities, key=lambda item: -item[1])
    
    return [word for word, _ in comp_similarities[:10]]  

```

```python
# test the function on know words:
results = vector_match('king','man','woman')
print(results)
```

```python
results = vector_match('dog','cat','tiguer')
print(results)
```

## Task #2: Perform VADER Sentiment Analysis on your own review
    Write code that returns a set of SentimentintensityAnalyzer polarity scores based on your own written review. 

```python
# Import SentimentIntensityAnalyzer and create an sid object
import nltk
nltk.download('vader_lexicon')  
from nltk.sentiment.vader import SentimentIntensityAnalyzer  
sid = SentimentIntensityAnalyzer()
```

```python
# Write reviews as a list of strings
my_reviews = [
    "The movie was excellent, but I didn't like the ending. It made me sad!",
    "The movie was great, I can't wait to see the second part.",
    "I didn't like the movie at all.",
    "I love this movie"
]
```

```python
# Obtain the sentiment scores for each review
for review in my_reviews:
    print(sid.polarity_scores(review))
```
